```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
metadata = pd.read_csv("drone3/metadata.csv")
print(metadata.info())
print(metadata.head())
```

```python
sns.set_theme(style="whitegrid")
plt.rcParams["figure.figsize"] = (14, 8)
plt.rcParams["axes.titlesize"] = 15
plt.rcParams["axes.labelsize"] = 13
plt.rcParams["xtick.labelsize"] = 12
plt.rcParams["ytick.labelsize"] = 12

plt.figure(figsize=(8, 8))
condition_metrics = metadata['condition'].value_counts()
condition_metrics
sns.barplot(
    x=condition_metrics.index, 
    y=condition_metrics.values, 
    hue=condition_metrics.index,
    palette="mako",
    legend=False
)
plt.xlabel("Погодные условия")
plt.ylabel('Количество фото')
plt.show()
```

```python
drone_type = metadata['drone_types'].str.split(';').explode()
drone_type = drone_type.value_counts()
sns.barplot(
    x = drone_type.values,
    y = drone_type.index, 
    palette='viridis', 
    hue = drone_type.index,
    legend=False
)
plt.xlabel('Кол-во дронов')
plt.ylabel('Тип дронов')
plt.show()
```
