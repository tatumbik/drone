```python
import os
from ultralytics import YOLO
from pathlib import Path

# ==================== НАСТРОЙКИ ====================
DATASET_YAML = "drone3/dataset.yaml"   # путь к dataset.yaml
MODEL_NAME   = "yolov8n.pt"              # можно заменить на yolov8s.pt / yolov8m.pt
EPOCHS       = 10
IMG_SIZE     = 320
BATCH_SIZE   = 16
PROJECT_DIR  = "runs/drone_training"     # папка для сохранения результатов
RUN_NAME     = "yolov8n_drone_multiclass"
# ===================================================

def main():
    # Проверяем наличие конфигурационного файла
    if not Path(DATASET_YAML).exists():
        raise FileNotFoundError(
            f"Не найден {DATASET_YAML}. Убедитесь, что датасет распакован "
            "и путь к dataset.yaml указан верно."
        )

    # Загружаем предобученную модель (transfer learning)
    model = YOLO(MODEL_NAME)

    # ---- ОБУЧЕНИЕ ----
    results = model.train(
        data=DATASET_YAML,
        epochs=EPOCHS,
        imgsz=IMG_SIZE,
        batch=BATCH_SIZE,
        project=PROJECT_DIR,
        name=RUN_NAME,
        exist_ok=True,
        # Дополнительные полезные параметры:
        patience=10,          # ранняя остановка, если метрики не улучшаются 30 эпох
        save=True,
        plots=True,           # сохранять графики (PR-кривые, confusion matrix и т.д.)
        verbose=True,
        device='cpu'     
    )

    # ---- ВАЛИДАЦИЯ НА ТЕСТОВОЙ ВЫБОРКЕ ----
    # После обучения автоматически загружаем лучшие веса и считаем метрики
    best_weights = Path(PROJECT_DIR) / RUN_NAME / "weights" / "best.pt"
    if best_weights.exists():
        print(f"\n✅ Обучение завершено. Лучшие веса: {best_weights}")
        print("\n📊 Запуск валидации на тестовой выборке...")

        val_model = YOLO(str(best_weights))
        metrics = val_model.val(
            data=DATASET_YAML,
            split="test",        # используем тестовую часть датасета
            imgsz=IMG_SIZE,
            batch=BATCH_SIZE,
            plots=True,
        )

        # ---- ВЫВОД КЛЮЧЕВЫХ МЕТРИК ----
        print("\n" + "=" * 60)
        print("РЕЗУЛЬТАТЫ ВАЛИДАЦИИ (тестовая выборка)")
        print("=" * 60)
        print(f"Precision (точность):       {metrics.box.mp:.4f}")
        print(f"Recall (полнота):           {metrics.box.mr:.4f}")
        print(f"mAP@0.5:                    {metrics.box.map50:.4f}")
        print(f"mAP@0.5:0.95:               {metrics.box.map:.4f}")

        # F1-мера (гармоническое среднее precision и recall)
        if metrics.box.mp > 0 and metrics.box.mr > 0:
            f1 = 2 * (metrics.box.mp * metrics.box.mr) / (metrics.box.mp + metrics.box.mr)
            print(f"F1-score:                   {f1:.4f}")

        # Метрики по каждому классу
        print("\n📌 Метрики по классам:")
        class_names = val_model.names  # словарь {id: name}
        for i, name in class_names.items():
            # В разных версиях Ultralytics API поля могут немного отличаться,
            # поэтому используем защищённый доступ
            try:
                p = metrics.box.p[i] if hasattr(metrics.box, "p") else float("nan")
                r = metrics.box.r[i] if hasattr(metrics.box, "r") else float("nan")
                ap50 = metrics.box.ap50[i] if hasattr(metrics.box, "ap50") else float("nan")
                ap = metrics.box.ap[i] if hasattr(metrics.box, "ap") else float("nan")
                print(f"  {name:15s} | P: {p:.3f} | R: {r:.3f} | mAP@0.5: {ap50:.3f} | mAP@0.5:0.95: {ap:.3f}")
            except Exception:
                print(f"  {name}: метрики по классу недоступны (проверьте версию Ultralytics)")

        print("\n📁 Все графики (confusion matrix, PR-кривые, F1-кривая и др.) "
              f"сохранены в: {Path(PROJECT_DIR) / RUN_NAME}")
    else:
        print("⚠️ Веса best.pt не найдены. Проверьте процесс обучения.")

if __name__ == "__main__":
    main()
```

```python
import cv2
import time
from collections import defaultdict, Counter
from pathlib import Path
from ultralytics import YOLO

# ==================== НАСТРОЙКИ ====================
# Путь к входному видео (обязательно укажите свой!)
VIDEO_PATH = "drone_video3.mp4"

# Путь к обученным весам модели
MODEL_PATH = "runs/detect/runs/drone_training/yolov8n_drone_multiclass/weights/best.pt"

# Куда сохранить аннотированное видео.
# Если None — сохранится рядом с исходным с суффиксом "_annotated.mp4"
OUTPUT_PATH = None

# Порог уверенности (0..1). Ниже — больше детекций, но больше ложных срабатываний.
CONF_THRESHOLD = 0.35

# Порог IoU для NMS (0..1).
IOU_THRESHOLD = 0.45

# Трекер: "bytetrack.yaml" (быстрее) или "botsort.yaml" (точнее при перекрытиях)
TRACKER = "bytetrack.yaml"

# Показывать видео в реальном времени (окно OpenCV). False — только запись файла.
SHOW_VIDEO = True

# Клавиша для досрочной остановки (в окне OpenCV)
STOP_KEY = "q"
# ===================================================


def run_inference(
    video_path: str,
    model_path: str,
    output_path: str | None,
    conf: float,
    iou: float,
    tracker: str,
    show: bool,
    stop_key: str = "q",
):
    # ---- Проверка входных путей ----
    video_file = Path(video_path)
    if not video_file.exists():
        raise FileNotFoundError(
            f"Видео не найдено: {video_path}\n"
            "Откройте скрипт и укажите правильный путь в переменной VIDEO_PATH."
        )

    model_file = Path(model_path)
    if not model_file.exists():
        raise FileNotFoundError(
            f"Модель не найдена: {model_path}\n"
            "Сначала обучите модель (train_drone.py) или укажите правильный путь "
            "в переменной MODEL_PATH."
        )

    # ---- Загрузка модели ----
    model = YOLO(str(model_file))
    print(f"✅ Модель загружена: {model_file}")
    print(f"   Классы: {list(model.names.values())}")

    # ---- Открытие видео ----
    cap = cv2.VideoCapture(str(video_file))
    if not cap.isOpened():
        raise IOError(f"Не удалось открыть видео: {video_path}")

    width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
    height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))
    fps_in = cap.get(cv2.CAP_PROP_FPS) or 30.0
    total_frames = int(cap.get(cv2.CAP_PROP_FRAME_COUNT)) or 0

    # ---- Подготовка выходного файла ----
    if output_path is None:
        output_path = str(video_file.with_name(
            video_file.stem + "_annotated.mp4"
        ))
    fourcc = cv2.VideoWriter_fourcc(*"mp4v")
    writer = cv2.VideoWriter(output_path, fourcc, fps_in, (width, height))
    if not writer.isOpened():
        raise IOError(
            f"Не удалось создать выходной видеофайл: {output_path}\n"
            "Проверьте права на запись в папку или измените OUTPUT_PATH."
        )

    # ---- Счётчики для метрик ----
    frame_count = 0
    inference_times = []
    detection_counts = Counter()
    confidence_sums = defaultdict(float)
    confidence_counts = defaultdict(int)
    unique_track_ids = set()

    start_time = time.time()

    print(f"\n🎬 Обработка видео: {video_path}")
    print(f"   Разрешение: {width}x{height}, FPS: {fps_in:.1f}, кадров: {total_frames}")
    print(f"   Сохранение в: {output_path}")
    if show:
        print(f"   Нажмите '{stop_key}' в окне для остановки.\n")

    while True:
        ret, frame = cap.read()
        if not ret:
            break

        frame_count += 1
        t0 = time.perf_counter()

        # ---- Трекинг ----
        results = model.track(
            source=frame,
            conf=conf,
            iou=iou,
            tracker=tracker,
            persist=True,
            verbose=False,
        )

        t1 = time.perf_counter()
        inference_times.append((t1 - t0) * 1000)

        # ---- Отрисовка и статистика ----
        annotated_frame = frame.copy()
        if results and results[0].boxes is not None:
            boxes = results[0].boxes
            xyxy = boxes.xyxy.cpu().numpy()
            confs = boxes.conf.cpu().numpy()
            cls_ids = boxes.cls.cpu().numpy().astype(int)
            ids = boxes.id.cpu().numpy().astype(int) if boxes.id is not None else None

            for i in range(len(xyxy)):
                x1, y1, x2, y2 = map(int, xyxy[i])
                conf_val = float(confs[i])
                cls_id = int(cls_ids[i])
                cls_name = model.names[cls_id]

                detection_counts[cls_name] += 1
                confidence_sums[cls_name] += conf_val
                confidence_counts[cls_name] += 1

                if ids is not None:
                    unique_track_ids.add(int(ids[i]))

                # Bounding box
                color = (0, 255, 0)
                cv2.rectangle(annotated_frame, (x1, y1), (x2, y2), color, 2)

                # Подпись
                label = f"{cls_name} {conf_val:.2f}"
                if ids is not None:
                    label += f" ID:{int(ids[i])}"

                (tw, th), _ = cv2.getTextSize(
                    label, cv2.FONT_HERSHEY_SIMPLEX, 0.6, 2
                )
                cv2.rectangle(
                    annotated_frame, (x1, y1 - th - 6), (x1 + tw, y1), color, -1
                )
                cv2.putText(
                    annotated_frame, label, (x1, y1 - 4),
                    cv2.FONT_HERSHEY_SIMPLEX, 0.6, (0, 0, 0), 2, cv2.LINE_AA
                )

        # ---- Запись и показ ----
        writer.write(annotated_frame)
        if show:
            cv2.imshow("Drone Detection", annotated_frame)
            if cv2.waitKey(1) & 0xFF == ord(stop_key):
                print("⏹️ Остановлено пользователем.")
                break

    # ---- Завершение ----
    cap.release()
    writer.release()
    if show:
        cv2.destroyAllWindows()

    total_time = time.time() - start_time
    avg_fps = frame_count / total_time if total_time > 0 else 0

    # ==================== ВЫВОД МЕТРИК ====================
    print("\n" + "=" * 60)
    print("📊 СТАТИСТИКА ПО ВИДЕО")
    print("=" * 60)
    print(f"Обработано кадров:           {frame_count}")
    print(f"Общее время обработки:       {total_time:.2f} сек")
    print(f"Средний FPS (с учётом I/O):  {avg_fps:.2f}")
    if inference_times:
        avg_inf = sum(inference_times) / len(inference_times)
        print(f"Среднее время инференса:     {avg_inf:.1f} мс/кадр")
        print(f"Потенциальный FPS модели:    {1000/avg_inf:.1f}")

    print(f"\nУникальных треков (ID):      {len(unique_track_ids)}")

    print("\n📌 Детекции по классам (суммарно по кадрам):")
    if detection_counts:
        for cls_name, count in detection_counts.most_common():
            avg_conf = confidence_sums[cls_name] / confidence_counts[cls_name]
            print(f"  {cls_name:15s} | детекций: {count:6d} | "
                  f"средняя уверенность: {avg_conf:.3f}")
    else:
        print("  Детекций не найдено. Попробуйте снизить CONF_THRESHOLD.")

    print(f"\n💾 Аннотированное видео сохранено: {output_path}")

    # ---- Сохранение метрик в CSV ----
    csv_path = Path(output_path).with_suffix(".csv")
    with open(csv_path, "w", encoding="utf-8") as f:
        f.write("class,detections,avg_confidence\n")
        for cls_name in detection_counts:
            avg_conf = confidence_sums[cls_name] / confidence_counts[cls_name]
            f.write(f"{cls_name},{detection_counts[cls_name]},{avg_conf:.4f}\n")
        f.write(f"\nTotal frames,{frame_count}\n")
        f.write(f"Unique tracks,{len(unique_track_ids)}\n")
        f.write(f"Average FPS,{avg_fps:.2f}\n")
    print(f"📄 Метрики сохранены в CSV: {csv_path}")


if __name__ == "__main__":
    # Проверка, что пользователь не забыл указать видео
    if VIDEO_PATH.startswith("path/to/"):
        raise SystemExit(
            "⚠️  Откройте скрипт и укажите путь к видео в переменной VIDEO_PATH "
            "в блоке НАСТРОЙКИ в начале файла."
        )

    run_inference(
        video_path=VIDEO_PATH,
        model_path=MODEL_PATH,
        output_path=OUTPUT_PATH,
        conf=CONF_THRESHOLD,
        iou=IOU_THRESHOLD,
        tracker=TRACKER,
        show=SHOW_VIDEO,
        stop_key=STOP_KEY,
    )
```
