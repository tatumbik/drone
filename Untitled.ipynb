```python
"""
Скрипт для подготовки датасета VisioDECT к обучению YOLOv8 (Classification).
Стратегия: сценарное разделение для предотвращения утечки данных.

  Train : все изображения из cloudy
  Val   : часть sunny (VAL_RATIO_FROM_SUNNY)
  Test  : оставшийся sunny + все evening
"""

import shutil
import random
from pathlib import Path
from collections import defaultdict

# ============================================================
# ⚙️  НАСТРОЙКИ
# ============================================================

SOURCE_DIR = Path("drone2/VisioDECT Dataset Upload")            # ← путь к распакованному датасету
TARGET_DIR = Path("visiodect_classification")  # ← куда сохранить результат

# Какая доля sunny-изображений пойдёт в val (остальное — в test)
VAL_RATIO_FROM_SUNNY = 0.6   # 60% sunny → val, 40% sunny → test

# Сценарии, по которым разделяем
TRAIN_SCENARIO = "cloudy"
VAL_SCENARIO   = "sunny"
TEST_SCENARIO  = "evening"

# Случайность для воспроизводимости
random.seed(42)

# Расширения изображений
IMAGE_EXTENSIONS = {".jpg", ".jpeg", ".png", ".JPG", ".JPEG", ".PNG"}

# ============================================================
# 🔍  АВТООПРЕДЕЛЕНИЕ ПАПОК МОДЕЛЕЙ ДРОНОВ
# ============================================================

def detect_drone_models(source_dir: Path) -> list:
    """Возвращает список папок моделей дронов в корне датасета."""
    models = []
    for item in sorted(source_dir.iterdir()):
        if item.is_dir() and (item / "images").exists():
            models.append(item.name)
    return models


def get_images(folder: Path) -> list:
    """Возвращает список файлов-изображений в папке."""
    if not folder.exists():
        return []
    return [f for f in folder.iterdir() if f.suffix in IMAGE_EXTENSIONS and f.is_file()]


def copy_with_scenario_prefix(files: list, source_scenario: str, target_dir: Path) -> int:
    """Копирует файлы в целевую папку, добавляя префикс сценария."""
    target_dir.mkdir(parents=True, exist_ok=True)
    count = 0
    for f in files:
        new_name = f"{source_scenario}_{f.name}"
        dst = target_dir / new_name
        # Если вдруг такой файл уже есть — не перезаписываем, а добавляем суффикс
        if dst.exists():
            dst = target_dir / f"{source_scenario}_{count}_{f.name}"
        shutil.copy2(f, dst)
        count += 1
    return count


# ============================================================
# 🚀  ОСНОВНАЯ ЛОГИКА
# ============================================================

def main():
    if not SOURCE_DIR.exists():
        raise FileNotFoundError(f"Не найдена папка с датасетом: {SOURCE_DIR.resolve()}")

    models = detect_drone_models(SOURCE_DIR)
    if not models:
        raise RuntimeError(f"В {SOURCE_DIR} не найдено ни одной папки с подпапкой 'images/'")

    print(f"📦 Найдено {len(models)} моделей дронов: {', '.join(models)}\n")

    # Счётчики для итогового отчёта
    stats = defaultdict(lambda: defaultdict(int))

    for model in models:
        print(f"── Обработка класса: {model} ──")
        model_src = SOURCE_DIR / model / "images"

        # 1) TRAIN: весь cloudy
        cloudy_imgs = get_images(model_src / TRAIN_SCENARIO)
        n = copy_with_scenario_prefix(
            cloudy_imgs, TRAIN_SCENARIO, TARGET_DIR / "train" / model
        )
        stats[model]["train"] = n
        print(f"   ✅ train (cloudy):   {n} изобр.")

        # 2) VAL / TEST: делим sunny
        sunny_imgs = get_images(model_src / VAL_SCENARIO)
        random.shuffle(sunny_imgs)
        split_idx = int(len(sunny_imgs) * VAL_RATIO_FROM_SUNNY)

        sunny_val  = sunny_imgs[:split_idx]
        sunny_test = sunny_imgs[split_idx:]

        n = copy_with_scenario_prefix(
            sunny_val, VAL_SCENARIO, TARGET_DIR / "val" / model
        )
        stats[model]["val"] = n
        print(f"   ✅ val (sunny):      {n} изобр.  (из {len(sunny_imgs)} sunny)")

        # 3) TEST: остаток sunny + весь evening
        evening_imgs = get_images(model_src / TEST_SCENARIO)

        n_sunny = copy_with_scenario_prefix(
            sunny_test, VAL_SCENARIO, TARGET_DIR / "test" / model
        )
        n_evening = copy_with_scenario_prefix(
            evening_imgs, TEST_SCENARIO, TARGET_DIR / "test" / model
        )
        stats[model]["test"] = n_sunny + n_evening
        print(f"   ✅ test:             {n_sunny} sunny + {n_evening} evening = {n_sunny + n_evening}\n")

    # ============================================================
    # 📊  ИТОГОВЫЙ ОТЧЁТ
    # ============================================================
    print("=" * 62)
    print(f"{'Класс':<22}{'Train':>10}{'Val':>10}{'Test':>10}")
    print("-" * 62)
    total_train = total_val = total_test = 0
    for model in models:
        t, v, te = stats[model]["train"], stats[model]["val"], stats[model]["test"]
        total_train += t; total_val += v; total_test += te
        print(f"{model:<22}{t:>10}{v:>10}{te:>10}")
    print("-" * 62)
    print(f"{'ИТОГО':<22}{total_train:>10}{total_val:>10}{total_test:>10}")
    print("=" * 62)

    print(f"\n🎉 Готово! Датасет сохранён в: {TARGET_DIR.resolve()}")
    print(f"   Структура: {TARGET_DIR}/{{train,val,test}}/<имя_класса>/*.jpg\n")
    print("Теперь можно запускать обучение:")
    print(f"   model.train(data='{TARGET_DIR}', epochs=50, imgsz=224)")


if __name__ == "__main__":
    main()
```
